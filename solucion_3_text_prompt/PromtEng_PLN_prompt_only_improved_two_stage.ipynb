# MELD (dataset) + FLAN-T5 + Prompt Engineering

Este notebook:

1. Descarga y prepara **MELD** (diálogos de *Friends*) con una **ventana de contexto** por turno.
2. Construye un `input_text` que incluye el **contexto previo** y el **turno objetivo**.
3. Evalúa un modelo **instruction-tuned** (FLAN-T5) **sin fine-tuning**, usando *prompt engineering*:
   - few-shot balanceado,
   - definiciones de etiquetas,
   - calibración contextual,
   - ensemble de prompts,
   - y *label scoring* (puntuar cada etiqueta por *loss*).

Objetivo: mejorar **precisión / macro-F1** sin entrenar el modelo.


````python
import kagglehub
import os
import pandas as pd
````

````python
path = kagglehub.dataset_download("zaber666/meld-dataset")
print("Path to dataset files:", path)
````

````python
# Cargar CSVs de texto
train_path = os.path.join(path+'/MELD-RAW/MELD.Raw/train/', "train_sent_emo.csv")
dev_path   = os.path.join(path+'/MELD-RAW/MELD.Raw/', "dev_sent_emo.csv")
test_path  = os.path.join(path+'/MELD-RAW/MELD.Raw/', "test_sent_emo.csv")

train = pd.read_csv(train_path, encoding="utf-8")
dev   = pd.read_csv(dev_path, encoding="utf-8")
test  = pd.read_csv(test_path, encoding="utf-8")
````

````python
# Limpieza mínima de caracteres raros (apóstrofo 0x92)
def clean_text(s: str) -> str:
    if not isinstance(s, str):
        return s
    return s.replace('\x92', "")#

for df_ in (train, dev, test):
    df_["Utterance"] = df_["Utterance"].apply(clean_text)
````

````python
# Unificar splits y quedarnos con las columnas relevantes

##%%
df = pd.concat(
    [
        train.assign(split="train"),
        dev.assign(split="dev"),
        test.assign(split="test"),
    ],
    ignore_index=True,
)

df = df[
    [
        "Dialogue_ID",
        "Utterance_ID",
        "Speaker",
        "Utterance",
        "Emotion",
        "Sentiment",
        "split",
    ]
]

print(df.head())
print(df.shape)
````

````python
# Construir estructura de diálogos ordenados

##%%
dialogues = (
    df.groupby("Dialogue_ID")
      .apply(lambda x: x.sort_values("Utterance_ID").to_dict("records"))
      .to_dict()
)

# Ejemplo de diálogo
dialogues[0][:5]
````

````python
# Normalizar hablantes (SPEAKER_0, SPEAKER_1, ... por diálogo)

##%%
def normalize_speakers(dialogue):
    speakers = list({turn["Speaker"] for turn in dialogue})
    mapping = {spk: f"SPEAKER_{i}" for i, spk in enumerate(speakers)}
    for turn in dialogue:
        turn["Speaker_Norm"] = mapping[turn["Speaker"]]
    return dialogue

dialogues_norm = {
    d_id: normalize_speakers(dialog)
    for d_id, dialog in dialogues.items()
}

dialogues_norm[0][:5]
````

````python
# Construir filas con ventana de contexto (por ejemplo 2 turnos anteriores)

##%%
def build_context_rows(dialogue, window=4):
    rows = []
    dialogue = sorted(dialogue, key=lambda x: x["Utterance_ID"])
    for i, turn in enumerate(dialogue):
        ctx = dialogue[max(0, i - window):i]
        ctx_text = " ".join(
            f'{t["Speaker_Norm"]}: {t["Utterance"]}' for t in ctx
        )
        rows.append(
            {
                "dialogue_id": turn["Dialogue_ID"],
                "utt_id": turn["Utterance_ID"],
                "speaker": turn["Speaker_Norm"],
                "context": ctx_text,
                "text": turn["Utterance"],
                "emotion": turn["Emotion"],
                "sentiment": turn["Sentiment"],
                "split": turn["split"],
            }
        )
    return rows

all_rows = []
for d_id, dialog in dialogues_norm.items():
    all_rows.extend(build_context_rows(dialog, window=4))

df_model = pd.DataFrame(all_rows)
df_model.head()
````

````python
# Construir el campo final para el prompt: contexto + turno objetivo (con speaker)
# Nota: incluir el speaker del turno actual suele ayudar a modelos instruction-tuned.

##%%
df_model["context"] = df_model["context"].fillna("").astype(str)

# Si no hay contexto previo, marcamos explícitamente "None" para evitar prompts vacíos ambiguos
ctx = df_model["context"].str.strip()
ctx = ctx.mask(ctx.eq(""), "None")

df_model["input_text"] = (
    "Dialogue context (previous turns):\n"
    + ctx
    + "\n\nTarget turn to classify (emotion of THIS utterance):\n"
    + df_model["speaker"].astype(str).str.strip()
    + ": "
    + df_model["text"].astype(str).str.strip()
)

df_model[["dialogue_id", "utt_id", "speaker", "input_text", "emotion", "split"]].head(10)
````

````python
# Lista de emociones (clases) presentes en MELD

##%%
emotions = sorted(df_model["emotion"].unique())
print("Emociones:", emotions)

# Separar splits (útil para evaluar solo en test)
train_df = df_model[df_model["split"] == "train"].reset_index(drop=True)
dev_df   = df_model[df_model["split"] == "dev"].reset_index(drop=True)
test_df  = df_model[df_model["split"] == "test"].reset_index(drop=True)

print(train_df["emotion"].value_counts())
````

# Prompt engineering (mejorado)

En esta sección **no hacemos fine-tuning**. En su lugar, mejoramos el rendimiento con técnicas de *prompt engineering*:

1. **Modelo más grande** (por defecto `flan-t5-large`), con **carga en 8-bit/4-bit** si está disponible.
2. **Few-shot balanceado**, eligiendo ejemplos **cortos** por clase (reduce truncation y hace el prompt más “limpio”).
3. **Definiciones de etiquetas** (guía semántica para separar emociones parecidas).
4. **Contextual calibration** (reduce el sesgo a etiquetas frecuentes como `neutral`).
5. **Prompt ensembling** (varias plantillas de prompt y promedio de puntuaciones).

La predicción se hace con **label scoring** (se puntúa cada etiqueta candidata por *loss*; menor = mejor).


````python
import os
import random
import torch
import numpy as np
import torch.nn.functional as F
from tqdm.auto import tqdm
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# -----------------------------
# 0) Setup
# -----------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# Para GPU moderna, TF32 suele acelerar sin afectar demasiado
if device.type == "cuda":
    torch.backends.cuda.matmul.allow_tf32 = True

# -----------------------------
# 1) Modelo (más grande + cuantización si es posible)
# -----------------------------
# Con 16GB suele ir bien:
# - flan-t5-large en fp16
# - flan-t5-xl en 8-bit (si tienes bitsandbytes)
MODEL_CANDIDATES = [
    "google/flan-t5-xl",
    "google/flan-t5-large",
    "google/flan-t5-base",
]

# -----------------------------
# Cuantización (opcional)
# -----------------------------
# Para cuantización 8-bit/4-bit, Transformers suele requerir *accelerate* y *bitsandbytes*
# (porque usa device_map="auto"). Si no están disponibles, usamos fp16 normal en GPU.
HAS_ACCELERATE = False
try:
    import accelerate  # noqa: F401
    HAS_ACCELERATE = True
except Exception:
    HAS_ACCELERATE = False

HAS_BNB = False
try:
    import bitsandbytes as bnb  # noqa: F401
    HAS_BNB = True
except Exception:
    HAS_BNB = False

# Por defecto, intenta cuantización solo si el entorno lo soporta
PREFER_8BIT = HAS_ACCELERATE and HAS_BNB   # 8-bit (requiere accelerate+bitsandbytes)
PREFER_4BIT = False                        # 4-bit aún más ligero; actívalo si lo necesitas

if device.type == "cuda" and not (HAS_ACCELERATE and HAS_BNB):
    print("ℹ️ accelerate/bitsandbytes no detectados: se usará fp16 en GPU (sin device_map).")

def load_seq2seq_model(candidates=MODEL_CANDIDATES):
    last_err = None
    for name in candidates:
        try:
            tok = AutoTokenizer.from_pretrained(name)
            kwargs = {}

            # Intento de cuantización (solo si CUDA)
            qconf = None
            if device.type == "cuda" and (PREFER_8BIT or PREFER_4BIT) and HAS_ACCELERATE and HAS_BNB:
                try:
                    from transformers import BitsAndBytesConfig
                    if PREFER_4BIT:
                        qconf = BitsAndBytesConfig(
                            load_in_4bit=True,
                            bnb_4bit_use_double_quant=True,
                            bnb_4bit_quant_type="nf4",
                            bnb_4bit_compute_dtype=torch.float16,
                        )
                    elif PREFER_8BIT:
                        qconf = BitsAndBytesConfig(load_in_8bit=True)
                except Exception:
                    qconf = None

            if qconf is not None:
                # Requiere accelerate + bitsandbytes
                kwargs["device_map"] = "auto"
                kwargs["quantization_config"] = qconf
            else:
                # fp16 en GPU
                if device.type == "cuda":
                    kwargs["torch_dtype"] = torch.float16

            model = AutoModelForSeq2SeqLM.from_pretrained(name, **kwargs)
            if qconf is None:
                model = model.to(device)

            model.eval()
            print(f"✅ Cargado modelo: {name} (quantized={qconf is not None})")
            return tok, model, name
        except Exception as e:
            last_err = e
            print(f"⚠️ No se pudo cargar {name}: {type(e).__name__}: {e}")
            continue
    raise RuntimeError(f"No se pudo cargar ningún modelo. Último error: {last_err}")

tok, model, MODEL_NAME = load_seq2seq_model()

# Mantén el final del prompt si hay truncation (lo más importante suele estar al final)
tok.truncation_side = "left"
MAX_INPUT_LEN = 512
SCORING_BATCH_SIZE = 8  # baja a 4 si hay OOM

# -----------------------------
# 2) Etiquetas + definiciones (guía semántica)
# -----------------------------
LABELS = [e.lower() for e in emotions]
LABELS_STR = ", ".join(LABELS)

LABEL_DEFS = {
    "anger":   "angry, irritated, hostile, blaming, shouting, frustrated",
    "disgust": "disapproval, repulsion, contempt, 'gross', 'ew', 'that\'s awful'",
    "fear":    "scared, anxious, worried, panic, intimidated",
    "joy":     "happy, excited, pleased, amused, laughing",
    "neutral": "no strong emotion; matter-of-fact; informational; calm",
    "sadness": "sad, disappointed, down, regretful, grieving",
    "surprise":"surprised, shocked, amazed, unexpected reaction",
}

# Verbalizers: múltiples formas de expresar cada etiqueta.
# (Suele ayudar con clases raras como disgust/fear y reduce errores por tokenización.)
VERBALIZERS = {
    "anger":   ["anger", "angry"],
    "disgust": ["disgust", "disgusted", "revulsion"],
    "fear":    ["fear", "afraid", "scared", "anxiety"],
    "joy":     ["joy", "happy", "excited", "amused"],
    "neutral": ["neutral", "no emotion", "none"],
    "sadness": ["sadness", "sad", "unhappy", "down"],
    "surprise":["surprise", "surprised", "shocked", "amazed"],
}

DEFS_BLOCK = "\n".join([f"- {k}: {LABEL_DEFS.get(k,'')}" for k in LABELS])

# -----------------------------
# 3) Few-shot balanceado (elige ejemplos cortos por clase)
# -----------------------------
SHOTS_PER_LABEL = 2      # sube a 3 si tu prompt no se trunca
MAX_EX_TOKENS = 120      # evita ejemplos demasiado largos

def _tok_len(s: str) -> int:
    return len(tok(s, add_special_tokens=False).input_ids)

def build_fewshot_block(df, labels, shots_per_label=SHOTS_PER_LABEL, seed=SEED):
    rnd = random.Random(seed)
    blocks = []

    for lab in labels:
        subset = df[df["emotion"].str.lower() == lab].copy()
        if subset.empty:
            continue

        # Calculamos longitud de tokens y nos quedamos con los más cortos
        subset["n_tok"] = subset["input_text"].astype(str).apply(_tok_len)
        short = subset[subset["n_tok"] <= MAX_EX_TOKENS]
        if short.empty:
            short = subset.sort_values("n_tok").head(200)
        else:
            short = short.sort_values("n_tok").head(300)

        # Muestreo para diversidad dentro de los más cortos
        take_n = min(shots_per_label, len(short))
        picked = short.sample(n=take_n, random_state=rnd.randint(0, 10**9))

        for _, row in picked.iterrows():
            x = str(row["input_text"]).strip()
            blocks.append(
                "Example:\n"
                "```\n" + x + "\n```\n"
                f"Label: {lab}\n"
            )

    rnd.shuffle(blocks)
    return "\n".join(blocks)

FEWSHOT = build_fewshot_block(train_df, labels=LABELS)

# Etiquetas para la etapa 2 (solo emociones, sin neutral)
LABELS_EMO = [l for l in LABELS if l != 'neutral']
LABELS_EMO_STR = ", ".join(LABELS_EMO)
FEWSHOT_EMO = build_fewshot_block(train_df, labels=LABELS_EMO)


# -----------------------------
# 4) Prompt ensembling + contextual calibration
# -----------------------------
def prompt_template_A(input_text: str) -> str:
    return (
        "Task: Classify the emotion expressed in the TARGET turn of a dialogue.\n"
        f"Choose exactly ONE label from: {LABELS_STR}\n"
        "Output ONLY the label (lowercase).\n\n"
        "Label definitions:\n"
        f"{DEFS_BLOCK}\n\n"
        f"{FEWSHOT}\n"
        "Now classify this target turn:\n"
        f"```\n{input_text.strip()}\n```\n"
        "Label:"
    )

def prompt_template_B(input_text: str) -> str:
    return (
        "You are an expert dialogue emotion classifier.\n"
        "Focus on the EMOTION of the target utterance, not the whole conversation.\n"
        f"Valid labels: {LABELS_STR}\n"
        "If there is no clear emotion, choose neutral.\n\n"
        f"{FEWSHOT}\n"
        "Dialogue turn:\n"
        f"{input_text.strip()}\n"
        "Answer with one label:\n"
        "Label:"
    )

def prompt_template_C(input_text: str) -> str:
    return (
        "Classify the emotion of the speaker in the TARGET turn.\n"
        f"Return one of: {LABELS_STR}\n\n"
        "Hints:\n"
        "- anger: irritation/blame\n"
        "- disgust: contempt/repulsion\n"
        "- fear: anxious/scared\n"
        "- joy: positive/excited\n"
        "- sadness: negative/down\n"
        "- surprise: shocked/unexpected\n"
        "- neutral: informational/calm\n\n"
        f"{FEWSHOT}\n"
        "TARGET:\n"
        f"{input_text.strip()}\n"
        "Label:"
    )

PROMPT_BUILDERS = [prompt_template_A, prompt_template_B, prompt_template_C]

# Prompt de calibración (content-free) para corregir prior de etiquetas
CALIBRATION_INPUT = (
    "Dialogue context (previous turns):\nNone\n\n"
    "Target turn to classify (emotion of THIS utterance):\n"
    "SPEAKER_0: [EMPTY]"
)

# -----------------------------
# 5) Label scoring batched + verbalizers + (opcional) 2 etapas
# -----------------------------
pad_id = tok.pad_token_id

def _compute_priors(df, labels, smooth=1e-3):
    vc = df["emotion"].str.lower().value_counts()
    pri = torch.tensor([float(vc.get(l, 0)) + smooth for l in labels], dtype=torch.float32)
    pri = pri / pri.sum()
    return pri

# Priors del dataset (train) para penalizar predicciones de clases raras
TRAIN_PRIORS_FULL = _compute_priors(train_df, LABELS)
TRAIN_PRIORS_EMO  = _compute_priors(train_df[train_df["emotion"].str.lower() != "neutral"], LABELS_EMO)

# Priors para etapa 1 (neutral vs emotional) calculados desde el train real
_neu_n = float((train_df["emotion"].str.lower() == "neutral").sum())
_emo_n = float((train_df["emotion"].str.lower() != "neutral").sum())
TRAIN_PRIORS_STAGE1 = torch.tensor([_neu_n + 1e-3, _emo_n + 1e-3], dtype=torch.float32)
TRAIN_PRIORS_STAGE1 = TRAIN_PRIORS_STAGE1 / TRAIN_PRIORS_STAGE1.sum()


# -----------------------------
# 5.1) Preparar candidatos (verbalizers) y scorer genérico
# -----------------------------
def _prepare_candidates(labels, verbalizers_dict):
    cand_texts = []
    label_to_cand_idxs = []
    for lab in labels:
        vs = verbalizers_dict.get(lab, [lab])
        idxs = list(range(len(cand_texts), len(cand_texts) + len(vs)))
        label_to_cand_idxs.append(idxs)
        cand_texts.extend(vs)

    cand_ids = tok(
        cand_texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
    ).input_ids.to(device)
    cand_dec_in = model.prepare_decoder_input_ids_from_labels(cand_ids)
    return cand_texts, label_to_cand_idxs, cand_ids, cand_dec_in

@torch.inference_mode()
def _score_candidates_batch(prompts, cand_ids, cand_dec_in, batch_size=SCORING_BATCH_SIZE):
    """Devuelve (len(prompts), C) con loss media por candidato."""
    all_losses = []
    C = cand_ids.size(0)

    for start in range(0, len(prompts), batch_size):
        chunk = prompts[start:start+batch_size]
        enc = tok(
            chunk,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=MAX_INPUT_LEN,
        ).to(device)

        P = enc.input_ids.size(0)

        # Repetimos prompts C veces para puntuar todos los candidatos en 1 forward
        input_ids = enc.input_ids.repeat_interleave(C, dim=0)
        attention_mask = enc.attention_mask.repeat_interleave(C, dim=0)

        decoder_input_ids = cand_dec_in.repeat(P, 1)
        label_ids = cand_ids.repeat(P, 1)

        out = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            decoder_input_ids=decoder_input_ids,
        )
        logits = out.logits  # (P*C, T, V)
        V = logits.size(-1)

        shift_logits = logits[:, :-1, :].contiguous()
        shift_labels = label_ids[:, 1:].contiguous()

        loss_tok = F.cross_entropy(
            shift_logits.view(-1, V),
            shift_labels.view(-1),
            reduction="none",
            ignore_index=pad_id,
        ).view(P*C, -1)

        denom = (shift_labels != pad_id).sum(dim=1).clamp(min=1)
        loss_per = loss_tok.sum(dim=1) / denom  # (P*C,)
        loss_per = loss_per.view(P, C)          # (P, C)

        all_losses.append(loss_per.detach().cpu())

    return torch.cat(all_losses, dim=0)

def _aggregate_by_label(cand_losses, label_to_cand_idxs):
    # cand_losses: (N, C) en CPU
    cols = []
    for idxs in label_to_cand_idxs:
        cols.append(torch.min(cand_losses[:, idxs], dim=1).values)
    return torch.stack(cols, dim=1)  # (N, L)

def make_label_scorer(labels, verbalizers_dict):
    _, label_to_idxs, cand_ids, cand_dec_in = _prepare_candidates(labels, verbalizers_dict)

    def scorer(prompts, batch_size=SCORING_BATCH_SIZE):
        cand_losses = _score_candidates_batch(prompts, cand_ids, cand_dec_in, batch_size=batch_size)
        return _aggregate_by_label(cand_losses, label_to_idxs)

    return scorer

# Scorers
score_full = make_label_scorer(LABELS, VERBALIZERS)
score_emo  = make_label_scorer(LABELS_EMO, VERBALIZERS)

# -----------------------------
# 5.2) Prompts para etapa 2 (solo emociones) + calibración
# -----------------------------
DEFS_BLOCK_EMO = "\n".join([f"- {k}: {LABEL_DEFS.get(k,'')}" for k in LABELS_EMO])

def prompt_template_A_emo(input_text: str) -> str:
    return (
        "Task: Classify the emotion expressed in the TARGET turn of a dialogue.\n"
        f"Choose exactly ONE label from: {LABELS_EMO_STR}\n"
        "Output ONLY the label (lowercase).\n\n"
        "Label definitions:\n"
        f"{DEFS_BLOCK_EMO}\n\n"
        f"{FEWSHOT_EMO}\n"
        "Now classify this target turn:\n"
        f"```\n{input_text.strip()}\n```\n"
        "Label:"
    )

def prompt_template_B_emo(input_text: str) -> str:
    return (
        "You are an expert dialogue emotion classifier.\n"
        f"Valid labels: {LABELS_EMO_STR}\n"
        "Pick the single best label for the TARGET turn.\n"
        "If unsure, pick the closest label.\n\n"
        "Definitions:\n"
        f"{DEFS_BLOCK_EMO}\n\n"
        f"{FEWSHOT_EMO}\n"
        "TARGET:\n"
        f"{input_text.strip()}\n"
        "Answer:"
    )

def prompt_template_C_emo(input_text: str) -> str:
    return (
        "Instruction: Determine the emotion of the TARGET utterance in the dialogue.\n"
        f"Return one of: {LABELS_EMO_STR}\n"
        "Return only the label, nothing else.\n\n"
        "Emotion guide:\n"
        f"{DEFS_BLOCK_EMO}\n\n"
        f"{FEWSHOT_EMO}\n"
        "Dialogue (context then target):\n"
        f"{input_text.strip()}\n"
        "Label:"
    )

PROMPT_BUILDERS_EMO = [prompt_template_A_emo, prompt_template_B_emo, prompt_template_C_emo]

def _get_calibration_losses(builders, scorer):
    calib = []
    for builder in builders:
        p = builder(CALIBRATION_INPUT)
        losses = scorer([p], batch_size=1)[0]  # (L,)
        calib.append(losses)
    return calib

CALIB_LOSSES_EMO = _get_calibration_losses(PROMPT_BUILDERS_EMO, score_emo)

# -----------------------------
# 5.3) Etapa 1 (neutral vs emotional) para subir recall de neutral
# -----------------------------
STAGE1_LABELS = ["neutral", "emotional"]
STAGE1_LABELS_STR = ", ".join(STAGE1_LABELS)

STAGE1_VERBALIZERS = {
    "neutral": ["neutral", "no emotion", "none"],
    "emotional": ["emotional", "has emotion", "non-neutral"],
}

def build_stage1_fewshot(df, n_neutral=5, n_emotional=3, seed=SEED):
    rnd = random.Random(seed)
    blocks = []

    neu = df[df["emotion"].str.lower() == "neutral"].copy()
    if len(neu) > 0:
        neu["toklen"] = neu["input_text"].apply(_tok_len)
        neu = neu.sort_values("toklen").head(n_neutral)
        for _, r in neu.iterrows():
            blocks.append(
                "Example:\n"
                f"```\n{r['input_text'].strip()}\n```\n"
                "Label: neutral\n"
            )

    emo = df[df["emotion"].str.lower() != "neutral"].copy()
    if len(emo) > 0:
        emo["toklen"] = emo["input_text"].apply(_tok_len)
        # Preferimos diversidad: muestreamos por emoción si se puede
        emo_blocks = []
        for lab in LABELS_EMO:
            sub = emo[emo["emotion"].str.lower() == lab].copy()
            if len(sub) == 0:
                continue
            sub = sub.sort_values("toklen").head(1)
            r = sub.iloc[0]
            emo_blocks.append(
                "Example:\n"
                f"```\n{r['input_text'].strip()}\n```\n"
                "Label: emotional\n"
            )
        if len(emo_blocks) < n_emotional:
            # Completa con ejemplos cortos adicionales
            emo = emo.sort_values("toklen").head(n_emotional)
            emo_blocks = [
                "Example:\n"
                f"```\n{r['input_text'].strip()}\n```\n"
                "Label: emotional\n"
                for _, r in emo.iterrows()
            ]
        blocks.extend(emo_blocks[:n_emotional])

    rnd.shuffle(blocks)
    return "\n".join(blocks)

FEWSHOT_STAGE1 = build_stage1_fewshot(train_df)

def prompt_stage1(input_text: str) -> str:
    return (
        "Task: Decide whether the TARGET turn expresses a CLEAR emotion.\n"
        "If the utterance is mostly factual, routine, polite, or the emotion is unclear, choose neutral.\n"
        f"Return one of: {STAGE1_LABELS_STR}\n\n"
        "Definitions:\n"
        "- neutral: no clear emotion; matter-of-fact; informational; calm\n"
        "- emotional: a clear emotion is expressed (e.g., anger, joy, fear, sadness, surprise, disgust)\n\n"
        f"{FEWSHOT_STAGE1}\n"
        "Now decide for this TARGET turn:\n"
        f"```\n{input_text.strip()}\n```\n"
        "Label:"
    )

PROMPT_BUILDERS_STAGE1 = [prompt_stage1]
score_stage1 = make_label_scorer(STAGE1_LABELS, STAGE1_VERBALIZERS)
CALIB_LOSSES_STAGE1 = _get_calibration_losses(PROMPT_BUILDERS_STAGE1, score_stage1)

# -----------------------------
# 5.4) Predicción (2 etapas)
# -----------------------------
USE_CALIBRATION = True

# Penaliza clases raras (reduce overprediction de fear/joy y suele subir recall de neutral)
PRIOR_PENALTY_STRENGTH_STAGE1 = 0.8
PRIOR_PENALTY_STRENGTH_STAGE2 = 0.6

# Margen: predice "emotional" solo si gana claramente (neutral es el default)
STAGE1_MARGIN = 0.15

@torch.inference_mode()
def predict_batch_two_stage(texts, batch_size=8, show_progress=True):
    preds = ["neutral"] * len(texts)

    it = range(0, len(texts), batch_size)
    if show_progress:
        it = tqdm(it, desc="Predict (2-stage)")

    for start in it:
        chunk_texts = texts[start:start+batch_size]

        # ---------- Stage 1: neutral vs emotional ----------
        s1_losses_ens = None
        for t_idx, builder in enumerate(PROMPT_BUILDERS_STAGE1):
            prompts = [builder(x) for x in chunk_texts]
            losses = score_stage1(prompts, batch_size=SCORING_BATCH_SIZE)  # (P,2) CPU

            if USE_CALIBRATION:
                losses = losses - CALIB_LOSSES_STAGE1[t_idx].unsqueeze(0)

            # prior penalty (dataset prior)
            pri = TRAIN_PRIORS_STAGE1
            losses = losses + PRIOR_PENALTY_STRENGTH_STAGE1 * (-torch.log(pri)).unsqueeze(0)

            s1_losses_ens = losses if s1_losses_ens is None else (s1_losses_ens + losses)

        s1_losses = s1_losses_ens / len(PROMPT_BUILDERS_STAGE1)

        # Decisión con margen (neutral default)
        # emotional si loss_emotional + margin < loss_neutral
        neutral_loss = s1_losses[:, 0]
        emotional_loss = s1_losses[:, 1]
        is_emotional = (emotional_loss + STAGE1_MARGIN < neutral_loss).tolist()

        # ---------- Stage 2: emociones (sin neutral) ----------
        if any(is_emotional):
            idxs = [i for i, flag in enumerate(is_emotional) if flag]
            emo_texts = [chunk_texts[i] for i in idxs]

            s2_losses_ens = None
            for t_idx, builder in enumerate(PROMPT_BUILDERS_EMO):
                prompts = [builder(x) for x in emo_texts]
                losses = score_emo(prompts, batch_size=SCORING_BATCH_SIZE)  # (Q,6) CPU

                if USE_CALIBRATION:
                    losses = losses - CALIB_LOSSES_EMO[t_idx].unsqueeze(0)

                # prior penalty (entre emociones, ayuda a disgust/fear vs joy)
                pri2 = TRAIN_PRIORS_EMO
                losses = losses + PRIOR_PENALTY_STRENGTH_STAGE2 * (-torch.log(pri2)).unsqueeze(0)

                s2_losses_ens = losses if s2_losses_ens is None else (s2_losses_ens + losses)

            s2_losses = s2_losses_ens / len(PROMPT_BUILDERS_EMO)
            best = torch.argmin(s2_losses, dim=1).tolist()
            emo_preds = [LABELS_EMO[i] for i in best]

            for local_i, global_i in enumerate(idxs):
                preds[start + global_i] = emo_preds[local_i]
        # else: se queda neutral

    return preds

# (Opcional) inspección rápida de un prompt
print("Ejemplo de prompt (Stage 1)\n---\n", prompt_stage1(train_df.loc[0, "input_text"])[:900], "\n---")
print("Ejemplo de prompt (Stage 2 - template A)\n---\n", prompt_template_A_emo(train_df.loc[0, "input_text"])[:900], "\n---")
````

````python
from sklearn.metrics import classification_report

# Evaluación en el conjunto de test (muestra)
sample_size = 2000
test_sample = test_df.sample(n=min(sample_size, len(test_df)), random_state=42)

texts = test_sample["input_text"].tolist()
y_true = test_sample["emotion"].str.lower().tolist()
y_pred = predict_batch_two_stage(texts)

print(
    classification_report(
        y_true,
        y_pred,
        labels=LABELS,
        target_names=LABELS,
        zero_division=0,
    )
)
````

## Notas / cómo exprimir más precisión (sin entrenar)

- Estos resultados mejoran bastante cuando **neutral** deja de estar infra-predicho.
  Por eso el notebook usa ahora un **clasificador en 2 etapas**:
  1) `neutral` vs `emotional`
  2) si `emotional`, clasifica entre `anger, disgust, fear, joy, sadness, surprise`
- Perillas útiles:
  - `STAGE1_MARGIN`: súbelo (0.15 → 0.25) para predecir *neutral* más a menudo.
  - `PRIOR_PENALTY_STRENGTH_STAGE1`: súbelo (0.8 → 1.2) para castigar más las clases raras en etapa 1.
  - `PRIOR_PENALTY_STRENGTH_STAGE2`: ajusta el balance entre emociones (0.4–0.9 suele ir bien).
  - `SHOTS_PER_LABEL`: 2 → 3 si tu prompt no se trunca.
  - `MAX_INPUT_LEN`: 512 suele ser buen compromiso; si trunca demasiado, sube a 768 (si cabe en VRAM).
- Si quieres **más velocidad**, reduce `PROMPT_BUILDERS_EMO` a 1 o 2 plantillas.
